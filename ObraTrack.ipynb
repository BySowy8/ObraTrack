```python
%%writefile modulo_historial.py
import pandas as pd
from IPython.display import display, HTML

def obtener_datos_bitacora(supabase_client):
    """
    Obtiene todos los registros de la tabla 'bitacora' desde Supabase.
    """
    try:
        response = supabase_client.table("bitacora").select("*").execute()
        data = response.data
        if not data:
            return pd.DataFrame()
        return pd.DataFrame(data)
    except Exception as e:
        print(f"Error al obtener los datos: {e}")
        return pd.DataFrame()

def renderizar_historial_html(df):
    """
    Renderiza el historial con diseño minimalista y diferencia visual para imprevistos.
    """
    if df.empty:
        display(HTML("""
            <div style="font-family: -apple-system, sans-serif; padding: 15px; border: 1px solid #e1e4e8; background: #f6f8fa; color: #586069; font-size: 13px; max-width: 900px;">
                [Aviso] No hay registros en la bitácora todavía.
            </div>
        """))
        return

    # Estilos CSS limpios y adaptados para imprevistos
    estilos = """
    <style>
        .obratrack-container {
            font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, Helvetica, Arial, sans-serif;
            margin: 20px 0;
            max-width: 900px;
        }
        .obratrack-title {
            font-size: 15px;
            font-weight: 600;
            color: #24292e;
            margin-bottom: 8px;
            border-bottom: 1px solid #eaeaea;
            padding-bottom: 6px;
        }
        .obratrack-table {
            width: 100%;
            border-collapse: collapse;
            font-size: 13px;
            color: #24292e;
            background: #ffffff;
            border: 1px solid #e1e4e8;
        }
        .obratrack-table th {
            background-color: #f6f8fa;
            color: #24292e;
            font-weight: 600;
            text-align: left;
            padding: 8px 12px;
            border-bottom: 1px solid #e1e4e8;
        }
        .obratrack-table td {
            padding: 8px 12px;
            border-bottom: 1px solid #eaeaea;
            vertical-align: top;
        }
        .obratrack-table tr:hover {
            background-color: #f8f9fa;
        }
        /* Estilo sutil para filas con imprevistos */
        .fila-imprevisto {
            background-color: #fffdfd;
        }
        .badge-imprevisto {
            display: inline-block;
            padding: 2px 6px;
            font-size: 11px;
            font-weight: 500;
            color: #d73a49;
            background-color: #ffeef0;
            border: 1px solid rgba(27,31,35,0.15);
            border-radius: 3px;
            margin-top: 4px;
        }
        .badge-normal {
            color: #28a745;
            font-size: 12px;
        }
    </style>
    """

    # Construimos las filas de la tabla manualmente para inyectar la lógica de imprevistos
    filas_html = ""
    for _, row in df.iterrows():
        fecha = row.get('fecha', 'Sin fecha')
        avance = row.get('descripcion_avance', 'Sin descripción')
        imprevisto = row.get('imprevisto', None)

        if imprevisto and str(imprevisto).strip() and str(imprevisto).lower() != 'none':
            # Si hay imprevisto
            estado_html = f"<span class='badge-imprevisto'> IMPREVISTO: {imprevisto}</span>"
            clase_fila = "fila-imprevisto"
        else:
            estado_html = "<span class='badge-normal'>✓ Normal</span>"
            clase_fila = ""

        filas_html += f"""
        <tr class="{clase_fila}">
            <td style="white-space: nowrap; width: 110px;">{fecha}</td>
            <td>{avance}</td>
            <td style="width: 250px;">{estado_html}</td>
        </tr>
        """

    html_final = f"""
    {estilos}
    <div class="obratrack-container">
        <div class="obratrack-title">Historial de Avances (Bitácora)</div>
        <table class="obratrack-table">
            <thead>
                <tr>
                    <th>Fecha</th>
                    <th>Avance del Día</th>
                    <th>Estado / Observación</th>
                </tr>
            </thead>
            <tbody>
                {filas_html}
            </tbody>
        </table>
    </div>
    """

    display(HTML(html_final))
```

```python
from datetime import date

class Bitacora:
    def __init__(self, supabase_client, tabla="Bitacora"):
        self.supabase = supabase_client
        self.tabla = tabla

    def registrar_avance(self, descripcion: str, fecha: str = None):
        """
        Registra un nuevo reporte diario de avance de obra.
        - descripcion: texto describiendo el avance del día
        - fecha: formato 'YYYY-MM-DD'. Si no se especifica, usa la fecha de hoy.
        """
        if fecha is None:
            fecha = date.today().isoformat()

        data = {
            "descripcion": descripcion,
            "fecha": fecha
        }

        response = self.supabase.table(self.tabla).insert(data).execute()
        print(f"✅ Bitácora registrada: {descripcion} ({fecha})")
        return response

    def obtener_bitacoras(self):
        """Devuelve todos los registros de bitácora, ordenados por fecha descendente."""
        response = self.supabase.table(self.tabla).select("*").order("fecha", desc=True).execute()
        return response.data

    def obtener_por_fecha(self, fecha: str):
        """Filtra los registros de bitácora por una fecha específica."""
        response = self.supabase.table(self.tabla).select("*").eq("fecha", fecha).execute()
        return response.data
```

```python
%%writefile modulo_login.py
import ipywidgets as widgets
from IPython.display import display, clear_output

class SistemaAuth:
    def __init__(self, supabase_client):
        self.supabase = supabase_client
        self.usuario_actual = None

    def autenticar(self, correo, contrasena):
        """Valida las credenciales en Supabase"""
        try:
            response = self.supabase.table("usuarios").select("*").eq("correo", correo).eq("contrasena", contrasena).execute()
            datos = response.data
            if datos and len(datos) > 0:
                self.usuario_actual = datos[0]
                return True, f"¡Bienvenido! Has iniciado sesión como: {self.usuario_actual['rol']}"
            else:
                return False, "Correo o contraseña incorrectos."
        except Exception as e:
            return False, f"Error de conexión: {e}"

    def registrar_usuario(self, correo, contrasena, rol):
        """Crea un nuevo usuario en la base de datos"""
        try:
            # 1. Validar que el correo no exista ya
            verificacion = self.supabase.table("usuarios").select("*").eq("correo", correo).execute()
            if verificacion.data and len(verificacion.data) > 0:
                return False, "Ese correo ya está registrado en el sistema."

            # 2. Insertar el nuevo usuario
            nuevo_usuario = {
                "correo": correo,
                "contrasena": contrasena,
                "rol": rol
            }
            # Nota: Asumimos que la columna 'id' en Supabase se genera automáticamente
            response = self.supabase.table("usuarios").insert(nuevo_usuario).execute()

            if response.data:
                return True, "✅ Usuario creado exitosamente. Ve a la pestaña 'Login' para ingresar."
            return False, "No se pudo registrar el usuario."
        except Exception as e:
            return False, f"Error de conexión: {e}"

def mostrar_interfaz_auth(auth_sistema):
    """Renderiza el formulario con pestañas para Login y Registro"""

    # ================= 1. PESTAÑA DE LOGIN =================
    titulo_login = widgets.HTML(value="<h3 style='color: #24292e;'>Iniciar Sesión</h3>")
    correo_login = widgets.Text(description="Correo:")
    pass_login = widgets.Password(description="Contraseña:")
    btn_login = widgets.Button(description="Ingresar", button_style='primary')
    salida_login = widgets.Output()

    def al_presionar_login(b):
        with salida_login:
            clear_output()
            correo = correo_login.value.strip()
            contrasena = pass_login.value.strip()

            if not correo or not contrasena:
                display(widgets.HTML("<div style='color: orange;'>⚠️ Llena todos los campos.</div>"))
                return

            btn_login.description = "Validando..."
            exito, mensaje = auth_sistema.autenticar(correo, contrasena)
            btn_login.description = "Ingresar"

            if exito:
                display(widgets.HTML(f"<div style='color: green; padding: 10px; background: #e6ffed; border: 1px solid #34d058; border-radius: 5px;'>✅ {mensaje}</div>"))
            else:
                display(widgets.HTML(f"<div style='color: red; padding: 10px; background: #ffeef0; border: 1px solid #d73a49; border-radius: 5px;'>❌ {mensaje}</div>"))

    btn_login.on_click(al_presionar_login)
    caja_login = widgets.VBox([titulo_login, correo_login, pass_login, btn_login, salida_login])


    # ================= 2. PESTAÑA DE REGISTRO =================
    titulo_reg = widgets.HTML(value="<h3 style='color: #24292e;'>Crear Nuevo Usuario</h3>")
    correo_reg = widgets.Text(description="Correo:")
    pass_reg = widgets.Password(description="Contraseña:")
    rol_reg = widgets.Dropdown(description="Rol:", options=['Contratista', 'Cliente'], value='Contratista')
    btn_reg = widgets.Button(description="Registrar", button_style='success')
    salida_reg = widgets.Output()

    def al_presionar_registro(b):
        with salida_reg:
            clear_output()
            correo = correo_reg.value.strip()
            contrasena = pass_reg.value.strip()
            rol = rol_reg.value

            if not correo or not contrasena:
                display(widgets.HTML("<div style='color: orange;'>⚠️ Llena todos los campos.</div>"))
                return

            btn_reg.description = "Guardando..."
            exito, mensaje = auth_sistema.registrar_usuario(correo, contrasena, rol)
            btn_reg.description = "Registrar"

            if exito:
                display(widgets.HTML(f"<div style='color: green;'>{mensaje}</div>"))
                correo_reg.value = ""
                pass_reg.value = ""
            else:
                display(widgets.HTML(f"<div style='color: red;'>❌ {mensaje}</div>"))

    btn_reg.on_click(al_presionar_registro)
    caja_registro = widgets.VBox([titulo_reg, correo_reg, pass_reg, rol_reg, btn_reg, salida_reg])


    # ================= 3. ENSAMBLAR PESTAÑAS =================
    pestanas = widgets.Tab(children=[caja_login, caja_registro])
    pestanas.set_title(0, 'Login')
    pestanas.set_title(1, 'Registro')

    display(pestanas)
```

```python
%%writefile modulo_obras.py
import ipywidgets as widgets
from IPython.display import display, clear_output
import datetime

class GestorObras:
    def __init__(self, supabase_client):
        self.supabase = supabase_client
        self.obra_activa = None  # guarda la obra elegida como "activa" en esta sesión

    def crear_obra(self, nombre, direccion, cliente, contratista_id, fecha_inicio, fecha_fin_estimada):
        """Valida y guarda una obra nueva en Supabase"""
        nombre = (nombre or "").strip()
        direccion = (direccion or "").strip()
        cliente = (cliente or "").strip()
        contratista_id = (contratista_id or "").strip()

        # ---- Validaciones (Manejo de Errores Visible) ----
        if len(nombre) < 3:
            return False, "El nombre de la obra debe tener al menos 3 caracteres."
        if len(direccion) < 5:
            return False, "La dirección debe tener al menos 5 caracteres."
        if len(cliente) < 3:
            return False, "El nombre del cliente debe tener al menos 3 caracteres."
        if not fecha_inicio:
            return False, "La fecha de inicio es obligatoria."
        if fecha_fin_estimada and fecha_fin_estimada < fecha_inicio:
            return False, "La fecha estimada de finalización no puede ser anterior a la fecha de inicio."

        try:
            nuevo_registro = {
                "nombre": nombre,
                "direccion": direccion,
                "cliente": cliente,
                "contratista_id": contratista_id or None,
                "fecha_inicio": fecha_inicio.isoformat(),
                "fecha_fin_estimada": fecha_fin_estimada.isoformat() if fecha_fin_estimada else None,
            }
            respuesta = self.supabase.table("obras").insert(nuevo_registro).execute()
            if respuesta.data:
                return True, f"✅ Obra '{respuesta.data[0]['nombre']}' guardada correctamente."
            return False, "No se pudo guardar la obra."
        except Exception as e:
            return False, f"Error de conexión: {e}"

    def listar_obras(self):
        """Trae todas las obras registradas, más recientes primero"""
        try:
            respuesta = self.supabase.table("obras").select("*").order("creado_en", desc=True).execute()
            return True, respuesta.data
        except Exception as e:
            return False, f"Error de conexión: {e}"

    def seleccionar_activa(self, obra):
        self.obra_activa = obra


def mostrar_interfaz_obras(gestor):
    """Renderiza el formulario de registro + la lista/selector de obra activa"""

    # ================= 1. FORMULARIO DE REGISTRO =================
    titulo_form = widgets.HTML(value="<h3 style='color: #24292e;'>Registrar obra nueva</h3>")
    nombre_in = widgets.Text(description="Nombre:", placeholder="Ej: Apto 302 - Remodelación cocina")
    direccion_in = widgets.Text(description="Dirección:", placeholder="Ej: Calle 45 #12-30")
    cliente_in = widgets.Text(description="Cliente:", placeholder="Ej: María Fernanda Ruiz")
    contratista_in = widgets.Text(description="Contratista:", placeholder="Tu nombre (temporal, sin login aún)")
    fecha_inicio_in = widgets.DatePicker(description="Inicio:", value=datetime.date.today())
    fecha_fin_in = widgets.DatePicker(description="Fin estimado:")
    btn_guardar = widgets.Button(description="Guardar obra", button_style="primary")
    salida_form = widgets.Output()

    def al_guardar(b):
        with salida_form:
            clear_output()
            btn_guardar.description = "Guardando..."
            exito, mensaje = gestor.crear_obra(
                nombre_in.value,
                direccion_in.value,
                cliente_in.value,
                contratista_in.value,
                fecha_inicio_in.value,
                fecha_fin_in.value,
            )
            btn_guardar.description = "Guardar obra"

            if exito:
                display(widgets.HTML(
                    f"<div style='color: green; padding: 10px; background: #e6ffed; "
                    f"border: 1px solid #34d058; border-radius: 5px;'>{mensaje}</div>"
                ))
                nombre_in.value = ""
                direccion_in.value = ""
                cliente_in.value = ""
                contratista_in.value = ""
                fecha_fin_in.value = None
                refrescar_lista()
            else:
                display(widgets.HTML(
                    f"<div style='color: red; padding: 10px; background: #ffeef0; "
                    f"border: 1px solid #d73a49; border-radius: 5px;'>⚠️ {mensaje}</div>"
                ))

    btn_guardar.on_click(al_guardar)
    caja_form = widgets.VBox([
        titulo_form, nombre_in, direccion_in, cliente_in, contratista_in,
        fecha_inicio_in, fecha_fin_in, btn_guardar, salida_form,
    ])

    # ================= 2. LISTA / SELECTOR DE OBRA ACTIVA =================
    titulo_lista = widgets.HTML(value="<h3 style='color: #24292e;'>Obras registradas</h3>")
    chip_activa = widgets.HTML(value="<i>Sin obra activa seleccionada</i>")
    salida_lista = widgets.Output()

    def actualizar_chip():
        if gestor.obra_activa:
            chip_activa.value = (
                f"<div style='color:#c64f1f; border:1px solid #c64f1f; padding:8px 12px; "
                f"border-radius:5px; display:inline-block;'>Obra activa: "
                f"<b>{gestor.obra_activa['nombre']}</b></div>"
            )
        else:
            chip_activa.value = "<i>Sin obra activa seleccionada</i>"

    def refrescar_lista(b=None):
        with salida_lista:
            clear_output()
            exito, obras = gestor.listar_obras()

            if not exito:
                display(widgets.HTML(f"<div style='color:red;'>⚠️ {obras}</div>"))
                return
            if not obras:
                display(widgets.HTML("<i>Todavía no hay obras registradas.</i>"))
                return

            for obra in obras:
                es_activa = gestor.obra_activa and gestor.obra_activa["id"] == obra["id"]
                borde = "#c64f1f" if es_activa else "#d0d7de"
                etiqueta = " · <b style='color:#c64f1f;'>(activa)</b>" if es_activa else ""

                info = widgets.HTML(
                    value=(
                        f"<div style='border:1px solid {borde}; padding:10px; border-radius:5px; margin-bottom:6px;'>"
                        f"<b>{obra['nombre']}</b>{etiqueta}<br>"
                        f"<span style='color:#57606a; font-size:13px;'>"
                        f"{obra.get('direccion') or 'Sin dirección'} · Cliente: {obra.get('cliente') or '—'}<br>"
                        f"Inicio: {obra.get('fecha_inicio') or '—'} · "
                        f"Fin estimado: {obra.get('fecha_fin_estimada') or '—'}"
                        f"</span></div>"
                    )
                )
                display(info)

                if not es_activa:
                    btn_elegir = widgets.Button(description=f"Trabajar en: {obra['nombre']}", layout=widgets.Layout(width="auto"))

                    def hacer_callback(obra_seleccionada):
                        def callback(b):
                            gestor.seleccionar_activa(obra_seleccionada)
                            actualizar_chip()
                            refrescar_lista()
                        return callback

                    btn_elegir.on_click(hacer_callback(obra))
                    display(btn_elegir)

    btn_refrescar = widgets.Button(description="🔄 Refrescar lista")
    btn_refrescar.on_click(refrescar_lista)

    caja_lista = widgets.VBox([titulo_lista, chip_activa, btn_refrescar, salida_lista])

    # ================= 3. MOSTRAR TODO =================
    display(widgets.HBox([caja_form, caja_lista]))
    refrescar_lista()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

class GestorCambios:
    def __init__(self, supabase_client):
        self.supabase = supabase_client

    def crear_solicitud(self, obra, descripcion, costo_estimado, dias_adicionales):
        """Registra una solicitud de cambio extra, pendiente de aprobación del cliente"""
        descripcion = (descripcion or "").strip()

        # ---- Validaciones (Manejo de Errores Visible) ----
        if not obra:
            return False, "Selecciona la obra a la que aplica este cambio."
        if len(descripcion) < 5:
            return False, "Describe el cambio con al menos 5 caracteres."
        try:
            costo_estimado = float(costo_estimado)
        except (TypeError, ValueError):
            return False, "El costo estimado debe ser un número."
        if costo_estimado < 0:
            return False, "El costo estimado no puede ser negativo."

        dias_val = None
        if dias_adicionales not in (None, ""):
            try:
                dias_val = int(dias_adicionales)
                if dias_val < 0:
                    return False, "Los días adicionales no pueden ser negativos."
            except (TypeError, ValueError):
                return False, "Los días adicionales deben ser un número entero."

        try:
            nuevo_registro = {
                "obra": obra,
                "descripcion": descripcion,
                "costo_estimado": costo_estimado,
                "dias_adicionales": dias_val,
                "estado": "pendiente",
            }
            respuesta = self.supabase.table("cambios_extra").insert(nuevo_registro).execute()
            if respuesta.data:
                return True, "✅ Solicitud de cambio registrada, pendiente de aprobación del cliente."
            return False, "No se pudo registrar la solicitud."
        except Exception as e:
            return False, f"Error de conexión: {e}"

    def listar_solicitudes(self, obra=None):
        """Trae las solicitudes de cambio, opcionalmente filtradas por obra"""
        try:
            consulta = self.supabase.table("cambios_extra").select("*").order("creado_en", desc=True)
            if obra:
                consulta = consulta.eq("obra", obra)
            respuesta = consulta.execute()
            return True, respuesta.data
        except Exception as e:
            return False, f"Error de conexión: {e}"

    def actualizar_estado(self, cambio_id, nuevo_estado):
        """El cliente aprueba o rechaza una solicitud"""
        try:
            respuesta = (
                self.supabase.table("cambios_extra")
                .update({"estado": nuevo_estado})
                .eq("id", cambio_id)
                .execute()
            )
            if respuesta.data:
                return True, f"Solicitud marcada como '{nuevo_estado}'."
            return False, "No se encontró la solicitud."
        except Exception as e:
            return False, f"Error de conexión: {e}"


def _obtener_lista_obras(supabase_client):
    """Reutiliza la tabla 'obras' ya existente, sin tocarla"""
    try:
        res = supabase_client.table("obras").select("nombre").execute()
        return [o["nombre"] for o in res.data if o.get("nombre")] or []
    except Exception:
        return []


def mostrar_interfaz_cambios(gestor):
    """Renderiza el formulario de solicitud + la lista de aprobación/rechazo"""

    obras_disponibles = _obtener_lista_obras(gestor.supabase)

    # ================= 1. FORMULARIO DE SOLICITUD =================
    titulo_form = widgets.HTML(value="<h3 style='color:#24292e;'>Solicitar cambio extra</h3>")
    obra_in = widgets.Dropdown(description="Obra:", options=obras_disponibles or ["(sin obras registradas)"])
    descripcion_in = widgets.Textarea(description="Cambio:", placeholder="Ej: Cambiar piso laminado por porcelanato en sala")
    costo_in = widgets.FloatText(description="Costo est. $:", value=0)
    dias_in = widgets.IntText(description="Días extra:", value=0)
    btn_guardar = widgets.Button(description="Enviar solicitud", button_style="primary")
    salida_form = widgets.Output()

    def al_guardar(b):
        with salida_form:
            clear_output()
            btn_guardar.description = "Enviando..."
            exito, mensaje = gestor.crear_solicitud(
                obra_in.value, descripcion_in.value, costo_in.value, dias_in.value
            )
            btn_guardar.description = "Enviar solicitud"

            color = ("green", "#e6ffed", "#34d058") if exito else ("red", "#ffeef0", "#d73a49")
            display(widgets.HTML(
                f"<div style='color:{color[0]}; padding:10px; background:{color[1]}; "
                f"border:1px solid {color[2]}; border-radius:5px;'>{mensaje}</div>"
            ))
            if exito:
                descripcion_in.value = ""
                costo_in.value = 0
                dias_in.value = 0
                refrescar_lista()

    btn_guardar.on_click(al_guardar)
    caja_form = widgets.VBox([titulo_form, obra_in, descripcion_in, costo_in, dias_in, btn_guardar, salida_form])

    # ================= 2. LISTA DE SOLICITUDES (aprobar / rechazar) =================
    titulo_lista = widgets.HTML(value="<h3 style='color:#24292e;'>Solicitudes registradas</h3>")
    salida_lista = widgets.Output()

    ESTILOS_ESTADO = {
        "pendiente": ("#b08800", "#fff8e6"),
        "aprobado": ("#16a34a", "#e6ffed"),
        "rechazado": ("#d73a49", "#ffeef0"),
    }

    def refrescar_lista(b=None):
        with salida_lista:
            clear_output()
            exito, solicitudes = gestor.listar_solicitudes()

            if not exito:
                display(widgets.HTML(f"<div style='color:red;'>⚠️ {solicitudes}</div>"))
                return
            if not solicitudes:
                display(widgets.HTML("<i>Todavía no hay solicitudes de cambio.</i>"))
                return

            for s in solicitudes:
                color_txt, color_fondo = ESTILOS_ESTADO.get(s["estado"], ("#57606a", "#f6f8fa"))
                dias_txt = f" · +{s['dias_adicionales']} días" if s.get("dias_adicionales") else ""

                display(widgets.HTML(
                    f"<div style='border:1px solid {color_txt}; background:{color_fondo}; "
                    f"padding:10px; border-radius:5px; margin-bottom:6px;'>"
                    f"<b>{s['obra']}</b> — {s['descripcion']}<br>"
                    f"<span style='font-size:13px;'>Costo estimado: ${s['costo_estimado']:.2f}{dias_txt} · "
                    f"Estado: <b style='color:{color_txt};'>{s['estado'].upper()}</b></span>"
                    f"</div>"
                ))

                if s["estado"] == "pendiente":
                    btn_aprobar = widgets.Button(description="✅ Aprobar (cliente)", button_style="success", layout=widgets.Layout(width="auto"))
                    btn_rechazar = widgets.Button(description="❌ Rechazar (cliente)", button_style="danger", layout=widgets.Layout(width="auto"))

                    def hacer_callback(sol_id, nuevo_estado):
                        def callback(b):
                            gestor.actualizar_estado(sol_id, nuevo_estado)
                            refrescar_lista()
                        return callback

                    btn_aprobar.on_click(hacer_callback(s["id"], "aprobado"))
                    btn_rechazar.on_click(hacer_callback(s["id"], "rechazado"))
                    display(widgets.HBox([btn_aprobar, btn_rechazar]))

    btn_refrescar = widgets.Button(description="🔄 Refrescar")
    btn_refrescar.on_click(refrescar_lista)

    caja_lista = widgets.VBox([titulo_lista, btn_refrescar, salida_lista])

    # ================= 3. MOSTRAR TODO =================
    display(widgets.HBox([caja_form, caja_lista]))
    refrescar_lista()
```

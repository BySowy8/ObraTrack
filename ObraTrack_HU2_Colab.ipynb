```python
# 5. Levantar el servidor de FastAPI en un hilo aparte y exponerlo con ngrok
import threading
import time

import uvicorn

from main import app

def correr_servidor():
    uvicorn.run(app, host="0.0.0.0", port=8000, log_level="info")

hilo_servidor = threading.Thread(target=correr_servidor, daemon=True)
hilo_servidor.start()
time.sleep(2)  # dar tiempo a que uvicorn arranque

tunel_publico = ngrok.connect(8000)
print(f"✅ API disponible en: {tunel_publico.public_url}")
print(f"📖 Documentación interactiva (Swagger): {tunel_publico.public_url}/docs")
```
